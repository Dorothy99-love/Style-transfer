```python
#Compare with Student with advisor
import pingouin as pg
import pandas as pd

# 1. Load the dataset
file_path = 'Prof vs stu.csv'
df_raw = pd.read_csv(file_path)

# 3. Transform to Long-format
# value_vars must match the EXACT strings in the CSV header
df = df_raw.melt(id_vars=['title'],
                 value_vars=['Student', 'Professional'],
                 var_name='judge',
                 value_name='rating')

# 4. Remove rows with missing values
df = df.dropna()

print("Columns created:", df.columns.tolist())
print(df.head())
```

```python
# 1. Calculate ICC
# All arguments (title, judge, rating) now match the columns from Cell 2 exactly
icc = pg.intraclass_corr(data=df, targets='title', raters='judge', ratings='rating')

# 2. Set 'Type' as index and display the results
icc_results = icc.set_index('Type')

print("--- Intraclass Correlation Coefficient (ICC) Results ---")
display(icc_results)
```

```python
import pandas as pd

# 1. Load the new dataset
file_path = 'G-Eval(GPT) intraclass corr.csv'
df_raw = pd.read_csv(file_path)

# 2. Transform to Long-format
# Note: var_name and value_name MUST match the targets/raters/ratings in Cell 3
df = df_raw.melt(id_vars=['title'],
                 value_vars=['Manual', 'G-Eval with gpt rate'],
                 var_name='judge',
                 value_name='rating')

# 3. Cleanup: Remove any empty rows
df = df.dropna()

print("Columns successfully created:", df.columns.tolist())
print(df.head())
```

```python
import pingouin as pg

# 1. Calculate ICC
# Targets are the items (title), raters are the judges (judge), ratings are the scores (rating)
icc = pg.intraclass_corr(data=df, targets='title', raters='judge', ratings='rating')

# 2. Set 'Type' as index for the final display
icc_display = icc.set_index('Type')

print("--- Intraclass Correlation Coefficient (ICC) Results ---")
display(icc_display)
```
